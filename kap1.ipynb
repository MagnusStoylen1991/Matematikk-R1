## Logaritme funksjoner

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.log10(x)

x=np.linspace(0,10,1000)

plt.plot(x,f(x))
plt.grid()

plt.axhline(color="black")
plt.axvline(color="red")

plt.show()
```

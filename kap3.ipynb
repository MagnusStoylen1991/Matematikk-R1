# Finne tangent linje


```python
import sympy as sp

x = sp.symbols("x")
f = 2*sp.log(x)

a = 1

m = sp.diff(f, x).subs(x, a)
b = f.subs(x, a)

tangent = sp.expand(b + m*(x-a))
print(tangent)
```

# Derivert

```python
import sympy as sp
x=sp.symbols('x')

f=x**2 + 3*x + 5

derivert=sp.diff(f,x)

print(derivert)
```
